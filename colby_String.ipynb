```python
import matplotlib.pyplot as plt
import numpy as np

L = 100
steps = L + 1
amp = 25
center = 50


x = np.linspace(0, L, steps)
y = np.zeros(len(x))

for i in range(len(x)):
    if (x[i]<=center):
        y[i] = ((amp)/(center))*(x[i])
    elif (x[i]>center):
        y[i] = (-(amp)/(L-center))*(x[i]-center)+amp

print(y)

        
plt.figure()
plt.plot(x, y)
plt.xlabel("String Length")
plt.ylabel("String Height")
plt.xlim(0, L)
plt.ylim(0, amp*2)
plt.grid()
```
