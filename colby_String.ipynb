```python
import matplotlib.pyplot as plt
import numpy as np

L = 100
steps = L + 1
amp = 25
center = 50
c = 1


x = np.linspace(0, L, steps)
y = np.zeros(len(x))

for i in range(len(x)):
    if (x[i]<=center):
        y[i] = ((amp)/(center))*(x[i])
    elif (x[i]>center):
        y[i] = (-(amp)/(L-center))*(x[i]-center)+amp

print(x)
print(y)
        
def stringfunc(x, y, c):
    a = np.zeros(len(x))
    for i in range(1, len(x)-1):
        a[i]=(c**2)*((y[i-1]-2*y[i]+y[i+1])/((x[i]-x[i-1])**2))
    
    
    

stringfunc(x,y,c)

        
plt.figure()
plt.plot(x, y)
plt.xlabel("String Length")
plt.ylabel("String Height")
plt.xlim(0, L)
plt.ylim(0, amp*2)
plt.grid()
```
